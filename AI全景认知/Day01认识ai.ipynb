| 概念     | 英文全称                           | 类型或位置      | 核心含义                  |
| ------ | ------------------------------ | ---------- | --------------------- |
| AI     | Artificial Intelligence        | 总体领域       | 让机器完成感知、判断、生成、规划等智能任务 |
| ML     | Machine Learning               | AI 的重要实现方法 | 让模型从数据或反馈中学习规律        |
| DL     | Deep Learning                  | ML 中的一类方法  | 使用多层神经网络学习复杂表示        |
| NLP    | Natural Language Processing    | 应用与研究方向    | 让计算机分析、理解、转换和生成人类语言   |
| 生成式 AI | Generative AI                  | 能力类型       | 根据输入生成文本、图片、音频、代码等新内容 |
| LLM    | Large Language Model           | 模型类型       | 处理和生成语言的大模型           |
| RAG    | Retrieval-Augmented Generation | 应用架构       | 检索相关资料，并让模型结合资料生成回答   |
| Agent  | Intelligent Agent / AI Agent   | 应用系统       | 围绕目标组织模型、工具、数据、状态和工作流 |

八个概念之间的主要关系可以概括为：

- AI 是总体领域，ML 是实现 AI 的重要方法，DL 是 ML 中的一类方法；
- NLP 是处理人类语言的方向，可以使用规则、ML、DL 或 LLM 实现；
- 生成式 AI 描述生成内容的能力，LLM 是处理和生成语言的模型类型；
- RAG 是组合检索系统与生成模型的应用架构，不是一种新的大模型；
- Agent 是组合模型、工具、数据、状态和工作流的应用系统，不是更大的语言模型；
- 同一个 AI 应用可以同时包含 NLP 任务，并使用规则、传统 ML、LLM、RAG 和 Agent 工作流。

| 分类维度      | 主要问题           | 本节概念              | 例子                          |
| --------- | -------------- | ----------------- | --------------------------- |
| 技术实现      | 它是用什么方法和架构实现的？ | ML、DL、Transformer | 使用深度神经网络，并采用 Transformer 架构 |
| 任务、模型与能力  | 它处理什么信息，能够做什么？ | NLP、LLM、生成式 AI    | 处理语言；生成文本、图片或音频             |
| 应用架构与系统形态 | 它怎样组织组件并完成任务？  | RAG、Agent         | 检索后生成；组合模型、工具、状态和工作流        |

**DL 和 LLM 的分类依据不同**。DL 按实现方法分类，LLM 按数据类型、任务对象和模型规模分类。不能仅凭名称把二者画成永远不变的严格同心圆；如果未来出现不依赖深度神经网络的大语言模型，它仍可能属于 LLM，却不属于 DL。

生成式 AI 强调系统能够生成新内容。文本生成、图像生成、语音生成都属于这个能力方向，它们可以使用不同的模型架构实现。

当前主流 LLM 通常具有文本生成能力，所以它们一般也是生成式 AI 的重要组成部分；但生成式 AI 不只有 LLM，LLM 也不只用于“写文章”，还可以用于分类、信息提取和评分等任务。

RAG 主要解决“怎样让生成模型使用外部知识”的问题，典型流程是检索资料、补充上下文、生成回答。Agent 主要解决“怎样围绕目标组织步骤并执行任务”的问题，可能把 RAG 作为其中一个知识组件。

Agent 强调围绕目标组织多个组件完成任务。它可能使用一个或多个 LLM，也可能结合传统 ML、规则引擎、搜索算法、数据库和业务 API。LLM 是 Agent 可以使用的组件，而 Agent 是把这些组件组织起来的完整系统。

以“智能客服工单系统”为例，同一个系统可以同时拥有多种标签：

- 工单优先级预测模型采用 **ML**；
- 图片凭证识别模型采用 **DL**；
- 回复生成模型是采用 **Transformer** 架构的 **LLM**；
- 理解工单内容和提取订单号属于 **NLP** 任务；
- 自动回复体现了**生成式 AI**能力；
- 检索售后政策并结合资料作答采用 **RAG** 架构；
- 查询订单、修改工单、记录结果的整体流程构成 **Agent**；
- 所有这些能力和系统都属于 **AI** 应用领域。

| 技术          | 可以做什么                         | 单独使用时的局限                 |
| ----------- | ----------------------------- | ------------------------ |
| Python 规则   | 用正则提取订单号，检查状态流转是否合法           | 用户换一种表达方式，规则可能失效         |
| 传统 ML       | 根据历史工单预测类别、紧急程度或升级风险          | 通常针对特定预测目标，不擅长自由生成完整回复   |
| DL / NLP 模型 | 理解更多样的文本表达，完成意图识别、情感分类或信息抽取   | 一个专用模型通常只解决一类任务          |
| LLM         | 理解问题、提取订单号、总结情绪并生成回复草稿        | 不知道当前真实物流状态，也可能生成错误信息    |
| RAG         | 检索当前退换货政策、发货时效与客服话术           | 能提供依据，但不会自动查询这个订单的实时状态   |
| 业务工具 / API  | 使用订单号查询真实订单与物流状态              | 只返回数据，不一定能理解用户意图或组织回复    |
| Agent 工作流   | 按顺序组织意图理解、工具查询、规范检索、回复生成和工单记录 | 系统更复杂，需要权限、超时、重试、审计和人工兜底 |

## 机器学习的四种常见学习方式

- **监督学习（Supervised Learning）**：有标准答案，模型根据“问题 + 答案”学习；
- **无监督学习（Unsupervised Learning）**：没有标准答案，模型从数据中发现结构和规律；
- **自监督学习（Self-supervised Learning）**：从原始数据中自动构造学习目标；
- **强化学习（Reinforcement Learning）**：通过行动获得奖励或惩罚，再逐步调整行动策略。

| 学习方式  | 训练数据有无人工标签           | 学习信号来源           | 典型任务        | 举例                            |
| ----- | -------------------- | ---------------- | ----------- | ----------------------------- |
| 监督学习  | 通常有明确标签，也可以由业务系统提供答案 | 将模型预测与正确答案进行比较   | 分类、回归       | 使用已标注优先级的历史工单，学习判断新工单的优先级     |
| 无监督学习 | 没有标签                 | 数据自身的相似性、距离和分布结构 | 聚类、降维       | 根据用户购买行为自动划分不同客户群体            |
| 自监督学习 | 不需要人工逐条标注            | 从原始数据中自动构造出来的答案  | 大模型预训练、表示学习 | 遮住一句话中的部分内容，或者根据前文预测下一个 Token |
| 强化学习  | 通常没有每一步的正确答案         | 与环境交互后获得的奖励或惩罚   | 游戏决策、机器人控制  | 游戏获胜得到正奖励，失败得到负奖励，逐步学习行动策略    |
