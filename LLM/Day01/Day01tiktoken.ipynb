### 3.3 第一次编码与解码

```python
import tiktoken

# o200k_base 是 tiktoken 内置的编码方案名称，不是模型名称。
# 它规定文本如何切分，以及 Token 与 Token ID 如何对应。
encoding = tiktoken.get_encoding("o200k_base")
text = "退款什么时候到账？"
# encode():接收字符串，返回按照当前编码方案切分得到的 Token ID 列表
token_ids = encoding.encode(text)
# decode():接收完整的 Token ID 列表，把它还原为字符串
decoded_text = encoding.decode(token_ids)
# 用完整 Token ID 序列解码，并与原文比较，检查编码和解码能否往返还原
print("原始文本：", text)
print("Token ID：", token_ids)
print("Token 数：", len(token_ids))
print("解码结果：", decoded_text)
print("是否还原：", decoded_text == text)
```

### 3.4 观察单个 Token 的原始字节

```python
import tiktoken


# 继续使用同一编码方案，避免因切分规则变化而得到不可比较的结果
encoding = tiktoken.get_encoding("o200k_base")
text = "收到啦，谢谢😊！"
token_ids = encoding.encode(text)   #文本->Token ID列表

for index, token_id in enumerate(token_ids, start=1):
    # 单个 Token 可能只包含某个字符的一部分字节，因此查看原始字节更可靠
    token_bytes = encoding.decode_single_token_bytes(token_id)
    print(index, token_id, token_bytes)
```

### 3.5 固定对比实验

```python
import tiktoken

# 使用同一个 `encoding` 比较三段固定文本
# 固定编码方案是对比实验的控制变量；换用其他编码方案，Token 数可能改变
encoding = tiktoken.get_encoding("o200k_base")

samples = {
    "中文工单": "退款什么时候到账？",
    "Python 代码": 'for item in orders: print(item["id"])',
    "聊天与表情": "收到啦，谢谢😊！",
}

for sample_name, text in samples.items():
    token_ids = encoding.encode(text)   #文本->编码

    # 字符数和 Token 数采用不同的计量规则，二者不一定相等
    print(sample_name, "字符数=", len(text), "Token 数=", len(token_ids))
```
